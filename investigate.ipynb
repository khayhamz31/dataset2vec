```python
import os
import openml

# === configuration ===
api_key = "c0d6200b271e73a8aec0904980876c3c"  # replace with your OpenML API key
openml.config.apikey = api_key

suite_id = 219  # change this to your benchmark suite ID
output_dir = "dataset_test"
os.makedirs(output_dir, exist_ok=True)

# === get datasets in suite ===
suite = openml.study.get_suite(suite_id)
dataset_ids = {openml.tasks.get_task(tid).dataset_id for tid in suite.tasks}
print(f"Found {len(dataset_ids)} datasets in suite {suite_id}")

# === download and save ===
for did in dataset_ids:
    print(f"Downloading dataset {did}...")
    dataset = openml.datasets.get_dataset(did)
    X, y, _, _ = dataset.get_data(target=dataset.default_target_attribute)
    save_path = os.path.join(output_dir, f"{did}.csv")
    X.assign(target=y).to_csv(save_path, index=False)
    print(f"Saved to {save_path}")

print("✅ All datasets downloaded.")
```
